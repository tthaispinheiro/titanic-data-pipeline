# Exploratory Data Analysis - Titanic Dataset

## Objetivo

Este notebook realiza uma análise exploratória do dataset Titanic para compreender a estrutura dos dados, identificar problemas de qualidade e analisar algumas características relacionadas à sobrevivência dos passageiros.

O tratamento e processamento dos dados são automatizados separadamente no arquivo `src/pipeline.py`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Carregamento dos dados

Nesta etapa, os dados brutos são carregados para análise.

```python
df = pd.read_csv("../data/raw/train.csv")
```

Primeiramente, vamos visualizar os primeiros registros do dataset.

```python
print(f"Quantidade de linhas: {df.shape[0]}")
print(f"Quantidade de colunas: {df.shape[1]}")
```

## 3. Análise da qualidade dos dados

Nesta etapa, são verificadas possíveis inconsistências, como valores ausentes e registros duplicados.

```python
df.isnull().sum()
```

```python
(df.isnull().sum() / len(df)) * 100
```

```python
missing_data =pd.DataFrame({
    'quantidade_nulos':df.isnull().sum(),
    'percentual_nulos':(df.isnull().sum() / len(df)) * 100
})

missing_data.sort_values(
    by='percentual_nulos',
    ascending=False
)

```

```python
duplicates = df.duplicated().sum()

print(f"Quantidade de registros duplicados: {duplicates}")
```

## 4. Análise das variáveis

Nesta seção, são analisadas algumas características dos passageiros e sua relação com a sobrevivência.

```python
df["Survived"].value_counts()
```

```python
survival_rate = df["Survived"].mean() * 100

print(f"Taxa geral de sobrevivência: {survival_rate:.2f}%")
```

### Sobrevivência por sexo

Vamos analisar se existe diferença na taxa de sobrevivência entre os diferentes sexos.

```python
df.groupby("Sex")["Survived"].mean()
```

```python
df.groupby("Sex")["Survived"].mean().plot(
    kind="bar",
    xlabel="Sex",
    ylabel="Survival Rate",
    title="Survival Rate by Sex"
)

plt.show()
```

### Sobrevivência por classe de passageiro

Nesta análise, verificamos a relação entre a classe do passageiro e a taxa de sobrevivência.

```python
df.groupby("Pclass")["Survived"].mean()
```

```python
df.groupby("Pclass")["Survived"].mean().plot(
    kind="bar",
    xlabel="Passenger Class",
    ylabel="Survival Rate",
    title="Survival Rate by Passenger Class"
)

plt.show()
```

### Distribuição das idades

Vamos analisar a distribuição das idades dos passageiros.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=30,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

### Análise dos valores ausentes

O dataset apresenta valores ausentes em algumas colunas. Essas informações foram utilizadas para definir as estratégias de tratamento implementadas posteriormente no pipeline.

```python
missing_columns = df.isnull().sum()

missing_columns = missing_columns[
    missing_columns > 0
].sort_values(ascending=False)

missing_columns
```

Durante a análise exploratória, foram identificados alguns pontos importantes:

- O dataset possui valores ausentes, principalmente nas colunas `Cabin` e `Age`.
- A coluna `Cabin` apresenta uma quantidade significativa de valores ausentes.
